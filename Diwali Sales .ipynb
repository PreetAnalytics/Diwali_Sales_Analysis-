```python
import pandas as pd 
import numpy as np 
import seaborn as sns 
import matplotlib.pyplot as plt 
```

```python
ds = pd.read_csv("C:/Users/harma/Desktop/Diwali Sales Data.csv")
```

```python
#DATA REDUCTION 
# Lets remove the unwanted rows of  Status and Unnamed1 
ds.drop(['Status', 'unnamed1'],axis=1,inplace=True)
```

```python
ds.head()
```

```python
ds[ds.isnull().any(axis=1)]
```

```python
ds.isnull().sum() .sum()
```

```python
# Removing all the rows that have null values 
ds.dropna(inplace=True)
```

```python
ds.isnull().sum()
```

```python
# Checking for data types of columns 
ds.dtypes
```

```python
# lets change the martial status from 1 to Yes and 0 to No 
ds["Marital_Status"] .replace({1:"Yes",0:"No"},inplace =True)
```

```python
ds.head()
```

```python
# FEATURE ENGINEERING 
ds["Age_Category"] = pd.cut(ds["Age"], bins=[0,18,35,50,100],labels=["Teen","Young_Adult","Middle_Aged","Senior"])
```

```python
ds.head()
```

```python
ds.groupby("Zone")["Amount"].sum().reset_index()

```

```python
# IDENTIFYING OUTLIERS 
Q1 = ds["Amount"].quantile(0.25)
Q3 = ds["Amount"].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = ds[(ds["Amount"] < lower_bound) | (ds["Amount"] > upper_bound)]
print(outliers)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Grouping data by Age Category and Gender, summing the Amount spent
age_gender_amount = ds.groupby(["Age_Category", "Gender"])["Amount"].sum().reset_index()

# Plotting the grouped bar graph
plt.figure(figsize=(10, 6))
sns.barplot(x="Age_Category", y="Amount", hue="Gender", data=age_gender_amount, palette="coolwarm")

# Customizing the plot
plt.xlabel("Age Category", fontsize=12)
plt.ylabel("Total Amount Spent", fontsize=12)
plt.title("Total Amount Spent by Age Category and Gender", fontsize=14)
plt.xticks(rotation=45)  # Rotate labels for better readability
plt.legend(title="Gender")  # Adding legend for Gender
plt.grid(axis="y", linestyle="--", alpha=0.7)

# Show the plot
plt.show()

```

```python
# Grouping data by Age Category and summing the Amount spent
State_amount = ds.groupby("State")["Amount"].sum().reset_index()

# Plotting the bar graph
plt.figure(figsize=(8,5))
sns.barplot(x="State", y="Amount", data=State_amount, palette="coolwarm")

# Customizing the plot
plt.xlabel("State", fontsize=10)
plt.ylabel("Total Amount Spent", fontsize=12)
plt.title("Total Amount Spent by State", fontsize=14)
plt.xticks(rotation=45)  # Rotate labels for better readability
plt.grid(axis="y", linestyle="--", alpha=0.7)

# Show the plot
plt.show()
```

```python
import matplotlib.pyplot as plt

# Grouping data by Zone and summing the Amount spent
zone_amount = ds.groupby("Zone")["Amount"].sum().reset_index()

# Plotting the pie chart
plt.figure(figsize=(8, 8))
plt.pie(
    zone_amount["Amount"], 
    labels=zone_amount["Zone"], 
    autopct="%1.1f%%", 
    colors=sns.color_palette("coolwarm", len(zone_amount)), 
    startangle=140, 
    wedgeprops={"edgecolor": "black"}
)

# Customizing the plot
plt.title("Total Amount Spent by Zone", fontsize=14)

# Show the plot
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Grouping data by Marital Status and summing the Amount spent
marital_amount = ds.groupby("Marital_Status")["Amount"].sum().reset_index()

# Plotting the bar graph
plt.figure(figsize=(8,5))
sns.barplot(x="Marital_Status", y="Amount", data=marital_amount, palette="coolwarm")

# Customizing the plot
plt.xlabel("Marital Status", fontsize=12)
plt.ylabel("Total Amount Spent", fontsize=12)
plt.title("Total Amount Spent by Marital Status", fontsize=14)
plt.grid(axis="y", linestyle="--", alpha=0.7)

# Show the plot
plt.show()
```
